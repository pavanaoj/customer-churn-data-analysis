# Customer Churn — Data Quality and Exploratory Analysis

**Objective:** Audit and clean a synthetic customer churn dataset, document quality decisions, and inspect initial patterns.

## 1. Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Load the dataset

```python
df = pd.read_csv("customer_churn_sample.csv")
print("Rows and columns:", df.shape)
df.head()
```

## 3. Profile columns and data types

```python
print(df.info())
```

## 4. Check missing values

```python
print(df.isnull().sum())
```

## 5. Check duplicates

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate CustomerID values:", df["CustomerID"].duplicated().sum())
```

## 6. Check numeric outliers using IQR

```python
numeric_cols = ["Age", "TenureMonths", "MonthlyCharges", "TotalCharges", "SupportTickets"]
for col in numeric_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(col, "outliers:", len(outliers))
```

## 7. Clean the data

```python
clean = df.copy()
for col in clean.select_dtypes(include="object").columns:
    clean[col] = clean[col].str.strip()
print("Cleaning completed. No values were invented or imputed.")
```

## 8. Summary statistics

```python
clean.describe()
```

## 9. Churn distribution

```python
print(clean["Churn"].value_counts())
clean["Churn"].value_counts().plot(kind="bar", title="Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```

## 10. Conclusion

The dataset has 15 records and 11 columns. No missing values, duplicate rows, duplicate CustomerID values, or IQR-based numeric outliers were found. Safe whitespace standardization was applied. No customer values were invented or randomly replaced. The dataset contains 7 churned customers and 8 non-churned customers.
